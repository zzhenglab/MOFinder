# Abstract triage input validation

Validate a small real-data subset without model requests. The example uses the first 12 reference papers and their associated bibliographic records. It checks input coverage and preserves the reference consensus labels.

## Inputs and setup

Install `python -m pip install -e ".[notebook]"` from the repository root, then run the cell below. The paths in this table are relative to `Demo/03_abstract_triage/`. All three files are included; no API key or PDF files are needed.

| Input | Location | Contents |
| --- | --- | --- |
| Bibliographic records, CSV | `metadata.csv` | `DOI`, title, source, keyword columns, and `Abstract`. |
| Human reference, CSV | `ground_truth.csv` | `DOI`, `Consensus GT`, and individual annotation columns. |
| Expected validation, JSON | `expected_validation.json` | Counts and validation results for this exact twelve-paper example. |

For another collection, place CSVs with the same schemas in a separate folder and change `EXAMPLE` below to that folder. Omit the expected-result comparison when using different inputs. The cell validates data in memory and writes no output files. Use [the main notebook](../../notebooks/01_abstract_triage.ipynb) for live screening or saved-prediction analysis.

Implementation: [input validation](../../src/mofinder/literature/triage.py). See the [source-to-code guide](../../docs/source_to_code.md) for the original workflow stages and their corresponding functions.


```python
from pathlib import Path
import json
from mofinder.literature.triage import validate_inputs, validation_summary

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "Demo/03_abstract_triage/metadata.csv").is_file())
EXAMPLE = ROOT / "Demo/03_abstract_triage"
validated = validate_inputs(EXAMPLE / "metadata.csv", EXAMPLE / "ground_truth.csv")
summary = validation_summary(validated)
expected = json.loads((EXAMPLE / "expected_validation.json").read_text())
assert summary == expected
summary
```

Expected: 12 reference papers, 12 scheduled abstracts, 9 Y and 3 N consensus labels, and zero missing abstracts.
